```python
N = int(input())
A = input()
value = A.split()
int_array = []

for i in range(N):
    int_array.append(int(value[i]))

index_min_value = int_array.index(min(int_array))
index_max_value = int_array.index(max(int_array))

int_array[index_min_value], int_array[index_max_value] = int_array[index_max_value], int_array[index_min_value]
print(*int_array)
```
