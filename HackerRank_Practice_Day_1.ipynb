Sensor Calibrator Min-Max Scalin

```python
numbers_input = input().split()

x = float(numbers_input[0])
min_v = float(numbers_input[1])
max_v = float(numbers_input[2])
norm = (x - min_v) / (max_v - min_v)
print(f"{norm:.2f}")
```

Mini Loss Tracker

```python
N = int(input())
target = float(input())
sum = 0

for i in range(N):
    input_number = float(input())
    sum += input_number

avg = sum / N
if(avg <= target):
    print("PASS")
else:
    print("RETRY")
```

Majority Vote Gate

```python
N = int(input())

yes_count = 0
no_count = 0

for i in range(N):
    input_decision = input().upper()
    if input_decision == "YES":
        yes_count += 1
    else:
        no_count += 1

if yes_count >= no_count:
    print("ACCEPT")
else:
    print("REJECT")
```

```python
numbers = input().split()
numbers_length = len(numbers)
sum = 0
for i in range(numbers_length):
    sum += int(numbers[i])

avg = sum / numbers_length

if avg < 85:
    print("Dark Image")
elif avg >=85 and avg <= 170:
    print("Normal Image")
elif avg > 170:
    print("Bright Image")
```

```python
message = input()

keywords = ["ai", "data", "model", "learn", "train", "neural"]

words = message.split()
count = 0

for keyword in keywords:
    if keyword in words:
        count += 1

if count >= 2:
    print("AI Detected")
else:
    print("Not AI Related")
```
