```python
inpt = input();
split_inpt = inpt.split();

numbers = [];
for i in range(len(split_inpt)):
    numbers.append(int(split_inpt[i]));

min = numbers[0];
max = numbers[0];

for i in range(len(numbers)):
    if numbers[i] < min:
        min = numbers[i];
    if numbers[i] > max:
        max = numbers[i];
print(min, max);
```
